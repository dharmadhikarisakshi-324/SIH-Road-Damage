```python
from collections import Counter
import xml.etree.ElementTree as ET

# ---------------------------------------------------------
# AUDIT RDD2022 INDIA TRAINING ANNOTATIONS
# ---------------------------------------------------------

class_counts = Counter()

total_objects = 0
images_with_damage = 0
negative_images = 0
invalid_xml = 0

annotation_files = list(TRAIN_ANNOTATIONS.glob("*.xml"))

for xml_file in annotation_files:

    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()

        objects = root.findall("object")

        if len(objects) == 0:
            negative_images += 1
        else:
            images_with_damage += 1

        for obj in objects:

            name = obj.find("name")

            if name is not None:
                class_name = name.text.strip()
                class_counts[class_name] += 1
                total_objects += 1

    except Exception as e:
        invalid_xml += 1
        print(f"Error reading {xml_file.name}: {e}")


# ---------------------------------------------------------
# RESULTS
# ---------------------------------------------------------

print("=" * 60)
print("RDD2022 INDIA — TRAINING DATASET AUDIT")
print("=" * 60)

print(f"Training images          : {len(list(TRAIN_IMAGES.glob('*')))}")
print(f"XML annotations          : {len(annotation_files)}")
print(f"Images with damage       : {images_with_damage}")
print(f"Negative images          : {negative_images}")
print(f"Total damage instances   : {total_objects}")
print(f"Invalid XML files        : {invalid_xml}")

print("\n" + "=" * 60)
print("CLASS DISTRIBUTION")
print("=" * 60)

for class_name, count in class_counts.most_common():

    percentage = (count / total_objects) * 100

    print(
        f"{class_name:<8} : "
        f"{count:>6} instances "
        f"({percentage:>6.2f}%)"
    )
```

```python
# ============================================================
# RDD2022 JAPAN — DATASET AUDIT
# ============================================================

from pathlib import Path
import xml.etree.ElementTree as ET
from collections import Counter

# Japan dataset path
JAPAN_ROOT = PROJECT_ROOT / "datasets" / "Japan"

JAPAN_TRAIN_IMAGES = JAPAN_ROOT / "train" / "images"
JAPAN_TRAIN_ANNOTATIONS = JAPAN_ROOT / "train" / "annotations" / "xmls"

# Check paths
print("=" * 60)
print("JAPAN DATASET PATHS")
print("=" * 60)

print("Japan root:", JAPAN_ROOT)
print("Train images:", JAPAN_TRAIN_IMAGES)
print("Train annotations:", JAPAN_TRAIN_ANNOTATIONS)

print("\nExists:")
print("Japan root:", JAPAN_ROOT.exists())
print("Images:", JAPAN_TRAIN_IMAGES.exists())
print("Annotations:", JAPAN_TRAIN_ANNOTATIONS.exists())


# ============================================================
# COUNT FILES
# ============================================================

japan_images = [
    f for f in JAPAN_TRAIN_IMAGES.iterdir()
    if f.is_file()
]

japan_xmls = list(JAPAN_TRAIN_ANNOTATIONS.glob("*.xml"))

print("\n" + "=" * 60)
print("JAPAN FILE COUNT")
print("=" * 60)

print("Train images      :", len(japan_images))
print("XML annotations   :", len(japan_xmls))


# ============================================================
# CLASS DISTRIBUTION
# ============================================================

japan_classes = Counter()

japan_total_objects = 0
japan_images_with_damage = 0
japan_negative_images = 0
japan_invalid_xml = 0

for xml_file in japan_xmls:

    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()

        objects = root.findall("object")

        if len(objects) > 0:
            japan_images_with_damage += 1
        else:
            japan_negative_images += 1

        for obj in objects:

            name = obj.find("name")

            if name is not None:
                class_name = name.text.strip()

                japan_classes[class_name] += 1
                japan_total_objects += 1

    except Exception as e:
        japan_invalid_xml += 1
        print("Error:", xml_file.name, e)


# ============================================================
# RESULTS
# ============================================================

print("\n" + "=" * 60)
print("JAPAN CLASS DISTRIBUTION")
print("=" * 60)

print("Images with damage :", japan_images_with_damage)
print("Negative images    :", japan_negative_images)
print("Total instances    :", japan_total_objects)
print("Invalid XML        :", japan_invalid_xml)

print("\nClasses:")

for class_name, count in japan_classes.most_common():

    percentage = (count / japan_total_objects) * 100

    print(
        f"{class_name:<8} : "
        f"{count:>6} instances "
        f"({percentage:>6.2f}%)"
    )
```
